```python
import pandas as pd

panel_A_h1_midas = pd.read_excel("Table_4A_PanelA_MIDAS.xlsx", index_col=0)
panel_A_h1_adl = pd.read_excel("Table_4A_PanelA_Multiplicative_MIDAS.xlsx", index_col=0)

panel_B_h4_midas = pd.read_excel("Table_4A_PanelB_MIDAS.xlsx", index_col=0)
panel_B_h4_adl = pd.read_excel("Table_4A_PanelB_Multiplicative_MIDAS.xlsx", index_col=0)

panel_C_h1_midas = pd.read_excel("Table_4B_PanelC_MIDAS.xlsx", index_col=0)
panel_C_h1_adl = pd.read_excel("Table_4B_PanelC_Multiplicative_MIDAS.xlsx", index_col=0)

panel_D_h4_midas = pd.read_excel("Table_4B_PanelD_MIDAS.xlsx", index_col=0)
panel_D_h4_adl = pd.read_excel("Table_4B_PanelD_Multiplicative_MIDAS.xlsx", index_col=0)
```

```python
panel_A = pd.concat(
    [
        panel_A_h1_midas,   # shape (5,4)
        panel_A_h1_adl,
        panel_B_h4_midas,
        panel_B_h4_adl
    ],
    axis=1,
    keys=[
        "Regular MIDAS",
        "Multiplicative MIDAS",
        "Regular MIDAS",
        "Multiplicative MIDAS"
    ]
)
panel_A = panel_A.round(3)

panel_A
```

```python
latex_panel_A = panel_A.to_latex(
    float_format="%.2f",
    multicolumn=True,
    multirow=True,
    escape=False
)
print(latex_panel_A)

```

```python
panel_B = pd.concat(
    [
        panel_C_h1_midas,   # shape (5,4)
        panel_C_h1_adl,
        panel_D_h4_midas,
        panel_D_h4_adl
    ],
    axis=1,
    keys=[
        "Regular MIDAS",
        "Multiplicative MIDAS",
        "Regular MIDAS",
        "Multiplicative MIDAS"
    ]
)
panel_B = panel_B.round(3)

panel_B
```
