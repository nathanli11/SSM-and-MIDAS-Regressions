```python
import pandas as pd
import numpy as np
```

# Import des datas

```python
file_path = 'data_qg.xlsx'

sheet_names = [
    'T10', 'T1', 'SP', 'IP', 'Emply',
    'Exptn', 'Exptn1', 'PI', 'LEI',
    'Manu', 'Manu1', 'Oil'
]


dfs = {
    name: pd.read_excel(file_path, sheet_name=name)
    for name in sheet_names
}

for name, df in dfs.items():
    df['Date'] = pd.to_datetime(df['Date'])
    value_col = df.columns[1]
    df = df.rename(columns={value_col: name})
    df = df.set_index('Date')
    dfs[name] = df[[name]]

data = pd.concat(dfs.values(), axis=1)
data = data.sort_index()
data['TERM'] = data['T10'] - data['T1']
data
```

```python
data['Manu_final'] = data['Manu'].combine_first(data['Manu1'])

data = data.drop(columns=['Manu', 'Manu1'])
data = data.rename(columns={'Manu_final': 'Manu'})
data
```

```python
data = data[data.index >= '1959-01-01']
data
```

```python
rates_m = data["TERM"].copy()
rates_m = rates_m.resample('M').mean()
rates_m.index = pd.to_datetime(rates_m.index)
rates_m
```

```python
monthly_vars = ['SP', 'IP', 'Emply', 'Emply', 'PI', 'LEI', 'Manu', 'Oil']
X_m = data[monthly_vars].copy()
X_m.index = pd.to_datetime(X_m.index)
X_m = X_m.resample('M').last()
X_m
```

```python
X_m = X_m.join(rates_m[['TERM']], how='inner')
X_m
```

```python
def log(x):
    return np.log(x)

def dlog(x):
    return np.log(x).diff()
```

```python
levels_var = ['TERM', 'LEI']
log_vars = ['Exptn']
dlog_vars = ['SP', 'IP', 'Emply', 'PI', 'Manu', 'Oil']

for v in log_vars:
    data[v] = log(data[v])

for v in dlog_vars:
    data[v] = dlog(data[v])
    
data
```
