```python
T10 = pd.read_excel('data_qg.xlsx', sheet_name='T10')
T1 = pd.read_excel('data_qg.xlsx', sheet_name='T1')
SP = pd.read_excel('data_qg.xlsx', sheet_name='SP')
IP = pd.read_excel('data_qg.xlsx', sheet_name='IP')
Emply = pd.read_excel('data_qg.xlsx', sheet_name='Emply')
Exptn = pd.read_excel('data_qg.xlsx', sheet_name='Exptn')
Exptn1 = pd.read_excel('data_qg.xlsx', sheet_name='Exptn1')
PI = pd.read_excel('data_qg.xlsx', sheet_name='PI')
LEI = pd.read_excel('data_qg.xlsx', sheet_name='LEI')
Manu = pd.read_excel('data_qg.xlsx', sheet_name='Manu')
Manu1 = pd.read_excel('data_qg.xlsx', sheet_name='Manu1')
Oil = pd.read_excel('data_qg.xlsx', sheet_name='Oil')
```

```python
file_path = 'data_qg.xlsx'

sheet_names = [
    'T10', 'T1', 'SP', 'IP', 'Emply',
    'Exptn', 'Exptn1', 'PI', 'LEI',
    'Manu', 'Manu1', 'Oil'
]


dfs = {
    name: pd.read_excel(file_path, sheet_name=name)
    for name in sheet_names
}

for name, df in dfs.items():
    df['observation_date'] = pd.to_datetime(df['observation_date'])
    value_col = df.columns[1]
    df = df.rename(columns={value_col: name})
    df = df.set_index('observation_date')
    dfs[name] = df[[name]]

data = pd.concat(dfs.values(), axis=1)
data = data.sort_index()

data
```
